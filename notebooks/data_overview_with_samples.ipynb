## Setup

```python
import os
import sys
import json
from pathlib import Path
from collections import Counter

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'src').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
    if not (PROJECT_ROOT / 'src').exists():
        raise FileNotFoundError('Could not find src/ directory')

sys.path.insert(0, str(PROJECT_ROOT / 'src'))
from sft import load_chunks

print(f'✓ Project: {PROJECT_ROOT.name}')
```

## 1. Asas Al-Balagha Dataset

```python
print('\n=== ASAS AL-BALAGHA (Classical Arabic) ===')

asas_path = PROJECT_ROOT / 'data' / 'processed' / 'chunks_asas_albalagha.jsonl'
asas_chunks = load_chunks(str(asas_path))

print(f'Total chunks: {len(asas_chunks):,}')
print(f'Dialect: Classical Arabic (from Asas Al-Balagha dictionary)')
print(f'\nSample chunks:')

for i in range(min(3, len(asas_chunks))):
    chunk = asas_chunks[i]
    text = chunk.get('chunk_text', '')[:200]
    print(f'\n{i+1}. {chunk.get("chunk_id")}')
    print(f'   {text}...')
```

## 2. Najdi Popular Dataset

```python
print('\n=== NAJDI POPULAR (Modern Saudi Dialect) ===')

najdi_path = PROJECT_ROOT / 'data' / 'processed' / 'chunks_najdi_popular.jsonl'

if najdi_path.exists():
    najdi_chunks = load_chunks(str(najdi_path))
    print(f'Total chunks: {len(najdi_chunks):,}')
    print(f'Dialect: Modern Najdi (Saudi popular words & expressions)')
    print(f'\nSample chunks:')
    
    for i in range(min(3, len(najdi_chunks))):
        chunk = najdi_chunks[i]
        text = chunk.get('chunk_text', '')[:200]
        print(f'\n{i+1}. {chunk.get("chunk_id")}')
        print(f'   {text}...')
else:
    print('⚠️  Najdi Popular dataset not found')
    najdi_chunks = []
```

## 3. SFT Samples (Single Dataset)

```python
print('\n=== SFT SAMPLES (Single Dataset - Asas Al-Balagha) ===')

sft_path = PROJECT_ROOT / 'data' / 'generated' / 'sft' / 'candidates.jsonl'

if sft_path.exists():
    sft_samples = []
    with open(sft_path, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            if i < 5:  # Load first 5 samples
                sft_samples.append(json.loads(line))
            else:
                break
    
    total_sft = sum(1 for _ in open(sft_path))
    print(f'Total SFT samples: {total_sft:,}')
    print(f'\nShowing {len(sft_samples)} samples:')
    
    for i, sample in enumerate(sft_samples, 1):
        print(f'\n{i}. {sample.get("sample_id")}')
        print(f'   Type: {sample.get("question_type")} | Difficulty: {sample.get("difficulty")}')
        print(f'   Q: {sample.get("question")}')
        thinking = sample.get('thinking_arabic', '')[:100]
        print(f'   T(AR): {thinking}...' if len(sample.get('thinking_arabic', '')) > 100 else f'   T(AR): {thinking}')
        print(f'   A: {sample.get("answer")[:100]}...')
else:
    print('⚠️  SFT samples not found')
```

## 4. DPO Pairs (Single Dataset)

```python
print('\n=== DPO PAIRS (Single Dataset - Asas Al-Balagha) ===')

dpo_path = PROJECT_ROOT / 'data' / 'generated' / 'dpo' / 'candidates.jsonl'

if dpo_path.exists():
    dpo_samples = []
    with open(dpo_path, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            if i < 5:  # Load first 5 pairs
                dpo_samples.append(json.loads(line))
            else:
                break
    
    total_dpo = sum(1 for _ in open(dpo_path))
    print(f'Total DPO pairs: {total_dpo:,}')
    print(f'\nShowing {len(dpo_samples)} pairs:')
    
    for i, pair in enumerate(dpo_samples, 1):
        prompt = pair.get('prompt', [{}])[0].get('content', '')[:80]
        chosen = pair.get('chosen', [{}])[0].get('content', '')[:80]
        rejected = pair.get('rejected', [{}])[0].get('content', '')[:80]
        
        print(f'\n{i}. {pair.get("pair_id")}')
        print(f'   Rejection: {pair.get("rejection_type")}')
        print(f'   Prompt:   {prompt}...')
        print(f'   Chosen:   {chosen}...')
        print(f'   Rejected: {rejected}...')
else:
    print('⚠️  DPO pairs not found')
```

## 5. Multi-Dataset Samples (If Generated)

```python
print('\n=== MULTI-DATASET SAMPLES (If Generated) ===')

sft_multi_path = PROJECT_ROOT / 'data' / 'generated' / 'sft' / 'candidates_multi.jsonl'
dpo_multi_path = PROJECT_ROOT / 'data' / 'generated' / 'dpo' / 'candidates_multi.jsonl'

if sft_multi_path.exists():
    sft_multi = []
    with open(sft_multi_path, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            if i < 3:
                sft_multi.append(json.loads(line))
            else:
                break
    
    total_sft_multi = sum(1 for _ in open(sft_multi_path))
    total_dpo_multi = sum(1 for _ in open(dpo_multi_path)) if dpo_multi_path.exists() else 0
    
    print(f'\n✓ Multi-dataset generation completed')
    print(f'  SFT Samples: {total_sft_multi:,}')
    print(f'  DPO Pairs: {total_dpo_multi:,}')
    print(f'\nSample from multi-dataset:')
    
    for i, sample in enumerate(sft_multi, 1):
        print(f'\n{i}. {sample.get("sample_id")}')
        print(f'   Type: {sample.get("question_type")} | Difficulty: {sample.get("difficulty")}')
        print(f'   Q: {sample.get("question")}')
else:
    print('\n⚠️  Multi-dataset not yet generated')
    print('Run: notebooks/multi_dataset_pipeline.ipynb')
```

## 6. Statistics & Comparison

```python
print('\n=== DATASET STATISTICS ===')
print(f"\n{'Dataset':<25s} {'Chunks':>10s} {'Type':>20s}")
print('─' * 60)

asas_count = len(asas_chunks) if asas_chunks else 0
najdi_count = len(najdi_chunks) if najdi_chunks else 0

print(f"{'Asas Al-Balagha':<25s} {asas_count:>10,d} {'Classical Arabic':>20s}")
if najdi_count > 0:
    print(f"{'Najdi Popular':<25s} {najdi_count:>10,d} {'Modern Dialect':>20s}")
    print(f"{'TOTAL':<25s} {asas_count + najdi_count:>10,d}")
else:
    print(f"{'Najdi Popular':<25s} {'N/A':>10s} {'Not Found':>20s}")

print('\n✓ Dataset overview complete!')
```
